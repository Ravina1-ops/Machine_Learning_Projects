import libraries

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn import linear_model 
```

Read CSV or get dataset

```python
df = pd.read_csv("canada per capita income data.csv")
df.head(5)
```

scatter the plot

```python
%matplotlib inline
plt.scatter(df['year'], df['per capita income(Us$)'], color = "purple", marker = "*")
plt.xlabel('year')
plt.ylabel('income per capita')
plt.title('Income Prrediction Scatter plot')
plt.show()
```

train the model using above data

```python
reg = linear_model.LinearRegression()
reg.fit(df[['year']], df['per capita income(Us$)'])
```

model is trained lets test using year 2022 prediction income


```python
p = reg.predict(pd.DataFrame({'year': [2022]}))
p
```

how it predict value lets cross check

```python
reg.coef_
```
