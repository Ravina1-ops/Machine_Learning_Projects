```python
import numpy as np
import pandas as pd
import math
from sklearn.linear_model import LinearRegression

# 1. Dataset Load Karo
df = pd.read_csv("test_scores.csv")
print(df.head())

# 2. Gradient Descent Function Define Karo
def gradient_descent(x, y):
    m_curr = 0
    b_curr = 0
    iterations = 1000000  # Max iterations
    n = len(x)
    learning_rate = 0.0002
    cost_previous = 0
    
    for i in range(iterations):
        # Prediction line: y = m*x + b
        y_predicted = m_curr * x + b_curr
        
        # Cost function (Mean Squared Error)
        cost = (1/n) * sum([val**2 for val in (y - y_predicted)])
        
        # Derivatives (m aur b ke respect mein)
        md = -(2/n) * sum(x * (y - y_predicted))
        bd = -(2/n) * sum(y - y_predicted)
        
        # Weights Update Step
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd
        
        # Stopping Condition: Agar cost mein change itna chhota ho jaye ki
        # previous cost aur current cost lagbhag same hon, toh loop break kar do
        if math.isclose(cost, cost_previous, rel_tol=1e-20):
            break
            
        cost_previous = cost
        
    print(f"Gradient Descent -> Iteration: {i}, Cost: {cost:.6f}, m: {m_curr:.5f}, b: {b_curr:.5f}")
    return m_curr, b_curr

# 3. Scikit-Learn se Compare Karne ke liye Function
def predict_using_sklearn(df):
    r = LinearRegression()
    r.fit(df[['math']], df.cs)
    return r.coef_[0], r.intercept_

# 4. Execution & Comparison
x = np.array(df.math)
y = np.array(df.cs)

# Custom Gradient Descent Run Karo
m_gd, b_gd = gradient_descent(x, y)

# Sklearn Run Karo
m_sk, b_sk = predict_using_sklearn(df)

# Output Verification
print("\n--- FINAL COMPARISON ---")
print(f"Gradient Descent : Coef (m) = {m_gd:.5f}, Intercept (b) = {b_gd:.5f}")
print(f"Scikit-Learn     : Coef (m) = {m_sk:.5f}, Intercept (b) = {b_sk:.5f}")
```
